# Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
import numpy as np
```

# River Water Level 2021-2025

```python
df_river_2021_2025=pd.read_csv(r"Datasets\\River Water Level\\TamilNadu_River_Water_Level_Hourly_Telemetry_2021-2025.csv")
df_river_2021_2025.head()
```

```python
df_river_2021_2025.drop(columns=['Unnamed: 0'],inplace=True, errors='ignore')
df_river_2021_2025.rename(columns={'River Water Level Telemetry Hourly (meter)':'Hourly Water Level'},inplace=True)
```

```python
df_river_2021_2025['Data Acquisition Time']=pd.to_datetime(df_river_2021_2025['Data Acquisition Time'],format='%d-%m-%Y %H:%M')
df_river_2021_2025.info()
```

```python
cols_to_drop = [
    'SlNo',
    'Agency',
    'State LGD Code',
    'State',
    'Basin',
    'Tributary',
    'Subtributary',
    'SubSubtributary',
    'Local River',
    'Is_DischargeDataAvailable',
    'RL_of_zeroGauge'
]

df_river_2021_2025.drop(columns=cols_to_drop, inplace=True, errors='ignore')
df_river_2021_2025.info()
```

# River Water Level 2026-2030

```python
df_river_2026_2030=pd.read_csv(r"Datasets\\River Water Level\\TamilNadu_River_Water_Level_Hourly_Telemetry_2026-2030.csv")
df_river_2026_2030.head()
```

```python
df_river_2026_2030.drop(columns=['Is_DischargeDataAvailable','RL_of_zeroGauge','MeanSeaLevel','SlNo','Agency','State','State LGD Code','River','Basin','Tributary','Subtributary','SubSubtributary','Local River'],inplace=True)
df_river_2026_2030.rename(columns={'River Water Level Telemetry Hourly (meter)':'Hourly Water Level'},inplace=True)
df_river_2026_2030['Data Acquisition Time']=pd.to_datetime(df_river_2026_2030['Data Acquisition Time'],format='%d-%m-%Y %H:%M')
df_river_2026_2030.head()
```

# River Water 2021-2030

```python
df_river_2021_2030 = pd.concat([df_river_2021_2025, df_river_2026_2030],ignore_index=True)
df_river_2021_2030.head()
```

```python
df_river_2021_2030['Data Acquisition Time'].describe()
```

```python
assert(df_river_2021_2030.shape[0] == df_river_2021_2025.shape[0]+df_river_2026_2030.shape[0])
```

```python
df_river_2021_2030.sort_values(by=['Data Acquisition Time','District LGD Code'],inplace=True)
```

# River Discharge 2021-2025

```python
df_discharge_2021_2025=pd.read_csv(r"Datasets/River Water Level/TamilNadu_River_Discharge_Hourly_Telemetry_2021-2025.csv")
df_discharge_2021_2025.head()
```

```python
df_discharge_2021_2025.drop(columns=['SlNo','Agency','State','State LGD Code','River','Basin','Tributary','Subtributary','SubSubtributary','Local River'],inplace=True)
df_discharge_2021_2025['Data Acquisition Time']=pd.to_datetime(df_discharge_2021_2025['Data Acquisition Time'],format='%d-%m-%Y %H:%M')
df_discharge_2021_2025.rename(columns={'Manual Daily River Water Discharge (m3/sec)':'Discharge Level'},inplace=True)
df_discharge_2021_2025.head()
```

# Final dataset

```python
river = (
    df_river_2021_2030
    .groupby(['Data Acquisition Time', 'District LGD Code','District'])
    ['Hourly Water Level']
    .mean()
    .reset_index()
)
river.to_csv("river_data.csv", index=False)
```
